outliers para enteros

```python
def detectar_outliers_iqr(datos, factor=1.5):
    """
    Calcula los outliers de cualquier estructura de datos numérica (Serie, Lista, Array)
    usando el método IQR.
    """
    # Convertir a Serie de Pandas y eliminar valores nulos para evitar errores
    s = pd.Series(datos).dropna()
    
    # Cálculo de cuartiles e IQR
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    
    # Límites
    lim_inf = q1 - (factor * iqr)
    lim_sup = q3 + (factor * iqr)
    
    # Identificar valores fuera de los límites
    es_outlier = (s < lim_inf) | (s > lim_sup)
    
    return {
        'total_outliers': es_outlier.sum(),
        'porcentaje': round((es_outlier.sum() / len(s)) * 100, 2),
        'limite_inferior': lim_inf,
        'limite_superior': lim_sup,
        'valores_outliers': s[es_outlier].tolist()
    }
```

```python
resultado = detectar_outliers_iqr(df['store'])
print(resultado)
```

outliers para texto 

```python
import pandas as pd

def outliers_longitud_texto(datos, factor=1.5):
    """Detecta textos anormalmente largos o cortos según la cantidad de caracteres."""
    s = pd.Series(datos).astype(str)
    longitudes = s.str.len()
    
    q1 = longitudes.quantile(0.25)
    q3 = longitudes.quantile(0.75)
    iqr = q3 - q1
    
    lim_inf = max(0, q1 - (factor * iqr))
    lim_sup = q3 + (factor * iqr)
    
    es_outlier = (longitudes < lim_inf) | (longitudes > lim_sup)
    
    return {
        'limite_inf_caracteres': lim_inf,
        'limite_sup_caracteres': lim_sup,
        'total_outliers': es_outlier.sum(),
        'ejemplos_outliers': s[es_outlier].tolist()
    }
```
